```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential,load_model
import pickle
```

```python
model=load_model("ann_model.h5")
with open("standardscaler.pkl","rb") as f:
				scaler=pickle.load(f)
with open("labelencoder.pkl","rb") as f:
				encoder=pickle.load(f)	
with open("onehotencoder.pkl","rb") as f:
				onehot=pickle.load(f)
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
df=pd.DataFrame([input_data])
df.head()
```

```python
# df["Geography"]=encoder.transform(df["Geography"])
df["Gender"]=encoder.transform(df["Gender"])
df.head()
```

```python
geo = onehot.transform([["France"]]).toarray()
```

```python
df=pd.concat([df,pd.DataFrame(geo,columns=onehot.get_feature_names_out())],axis=1)
df.drop("Geography",axis=1,inplace=True)
df.head()
```

```python
print(model.predict(scaler.transform(df)))
```
